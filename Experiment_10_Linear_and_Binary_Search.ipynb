# Experiment 10: Linear and Binary Search

## 1. Linear Search
Linear Search starts at the leftmost element of an array and compares the target with each element one by one.

- If the target matches an element, return its index.
- If the target does not match any element, return `-1`.

Linear Search works on both sorted and unsorted lists.

### Program 1: Linear Search
Write a function that searches for a target value in a list and returns its index, or `-1` if the value is not found.

````python
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

# Call the function
numbers = [10, 20, 30, 40, 50]
target = 30

result = linear_search(numbers, target)
if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")
````

**Expected output**
```text
Element found at index: 2
```

### Program 2: Linear Search for n Numbers
Ask the user how many elements to enter, read the elements, and search for a target. The position shown is a zero-based index, matching the original program.

````python
n = int(input("Enter the number of elements: "))
arr = []

for i in range(n):
    num = int(input(f"Enter element {i + 1}: "))
    arr.append(num)

target = int(input("Enter the element to search: "))
found = False

for i in range(n):
    if arr[i] == target:
        print("Element found at position:", i)
        found = True
        break

if not found:
    print("Element not found")
````

## 2. Binary Search
Binary Search is an efficient algorithm for searching a **sorted** array/list. It repeatedly divides the search interval in half and has time complexity **O(log N)**.

### How Binary Search works
1. Find the middle element.
2. If the middle element equals the target, return its index.
3. If the target is smaller, search the left half.
4. If the target is larger, search the right half.
5. Repeat until the target is found or the search interval is empty.

The supplied program sorts the entered list first, so the binary search can work correctly.

### Program 3: Binary Search
Read numbers from the user, sort the list, and use the binary search loop to locate the target.

````python
n = int(input("Enter the number of elements: "))
arr = []

for i in range(n):
    num = int(input(f"Enter element {i + 1}: "))
    arr.append(num)

# Sort the list before binary search
arr.sort()
print("Sorted list:", arr)

target = int(input("Enter the element to search: "))
low = 0
high = n - 1
found = False

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == target:
        print("Element found at position:", mid)
        found = True
        break
    elif target < arr[mid]:
        high = mid - 1
    else:
        low = mid + 1

if not found:
    print("Element not found")
````

- **Binary Search:** requires sorted data; worst-case time complexity is **O(log N)**.
