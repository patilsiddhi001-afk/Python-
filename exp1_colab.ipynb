```python
# Grade using if...elif
score = int(input("Enter the student's score: "))

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

```python
# Nested If Else Statement
a, b = 10, 20

if a < 50:
    if b < 100:
        print("a is less than 50 and b is less than 100")
    else:
        print("a is less than 50 but b is not less than 100")
else:
    print("a is not less than 50")
```

```python
# Largest of 3 numbers
num1 = int(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

if (num1 >= num2) and (num1 >= num3):
    largest = num1
elif (num2 >= num1) and (num2 >= num3):
    largest = num2
else:
    largest = num3

print("The largest number is", largest)
```

```python
# Iterating through a list
list1 = ['MIT', 'WPU', 'Pune']
for i in list1:
    print(i)
```

```python
# Break Statement
# The break statement terminates the loop immediately.
list1 = ['MIT', 'WPU', 'Pune']

for i in list1:
    if i == 'Pune':
        break
    print(i)
```

```python
# Continue Statement
# The continue statement skips the current iteration.
list1 = ['MIT', 'WPU', 'Pune', 'India']

for i in list1:
    if i == 'Pune':
        continue
    print(i)
```

```python
# Break inside a range loop
for i in range(5):
    if i == 3:
        break
    print(i)
```

```python
# Continue inside a range loop
for i in range(5):
    if i == 3:
        continue
    print(i)
```
