## Part 1 — Python Modules

A **module** is a Python file (`.py`) that contains reusable functions. We will create `module1.py` for prime checking and factorial calculation, and `module2.py` for arithmetic operations.

### Program 1 — Create `module1.py`
This module contains `check_prime(num)` and `calculate_factorial(num)`. The functions print their results.

```python
%%writefile module1.py
def check_prime(num):
    if num < 2:
        print("Not prime")
        return

    if num == 2:
        print("Prime number")
        return

    if num % 2 == 0:
        print("Not prime")
        return

    for count in range(3, int(num ** 0.5) + 1, 2):
        if num % count == 0:
            print("Not prime")
            return

    print("Prime number")


def calculate_factorial(num):
    if num < 0:
        print("Please enter a non-negative integer!")
        return

    fact = 1
    for count in range(1, num + 1):
        fact *= count

    print("Factorial of", num, "is", fact)
```

### Program 2 — Create `module2.py`
This module contains addition, subtraction, multiplication and division functions.

```python
%%writefile module2.py
def add(a, b):
    return a + b


def sub(a, b):
    return a - b


def mul(a, b):
    return a * b


def div(a, b):
    if b == 0:
        print("Error: Division by zero is not allowed.")
        return None
    return a / b
```

### Program 3 — Import and use functions from `module1.py`

```python
from module1 import check_prime, calculate_factorial

number = int(input("Enter a non-negative integer: "))
check_prime(number)
calculate_factorial(number)
```

### Program 4 — Import and use functions from `module2.py`

```python
import module2 as m

a, b = map(float, input("Enter two numbers separated by a space: ").split())

print("Addition:", m.add(a, b))
print("Subtraction:", m.sub(a, b))
print("Multiplication:", m.mul(a, b))
print("Division:", m.div(a, b))
```

## Part 4 — NumPy

NumPy is a library for scientific computing. Its `ndarray` object stores elements in an n-dimensional array. An array has a fixed size, and its elements generally share the same data type.

### Program 12 — Import NumPy

```python
import numpy as np

print("NumPy version:", np.__version__)
```

### Program 13 — Create an array using `np.arange(6)`

```python
a = np.arange(6)
print(a)
```

### Program 14 — Create a one-dimensional array

```python
a = np.array([2, 3, 4])
print(a)
```

### Program 15 — Create a two-dimensional array

```python
b = np.array([(1.5, 2, 3), (4, 5, 6)])
print(b)
```

### Program 16 — Create a complex-valued array

```python
c = np.array([(1.5, 2, 3), (4, 5, 6)], dtype=complex)
print(c)
```

### Program 17 — Create a 3×5 array and inspect its attributes

```python
a = np.arange(15).reshape(3, 5)

print("Array:\n", a)
print("Shape:", a.shape)
print("Number of dimensions:", a.ndim)
print("Data type name:", a.dtype.name)
print("Number of elements:", a.size)
print("Python type:", type(a))
print("Size of each element in bytes:", a.itemsize)
```

### Program 18 — Subtract NumPy arrays

```python
a = np.array([20, 30, 40, 50])
print("Array a:", a)

b = np.arange(4)
print("Array b:", b)

c = a - b
print("a - b:", c)
```

### Program 19 — Select the first two rows and alternate columns

```python
a = np.array([
    [-1, 2, 0, 4],
    [4, -0.5, 6, 0],
    [2.6, 0, 7, 8],
    [3, -7, 4, 2.0]
])

a = a[:2, ::2]
print("First 2 rows and alternate columns (columns 0 and 2):\n", a)
```
