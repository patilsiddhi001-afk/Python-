## Data Structures
A **data structure** organizes and stores data so it can be accessed and processed efficiently.

Data structures can broadly be classified into:

- **Linear:** elements are arranged sequentially, one after another. Examples: arrays, lists, stacks, queues, and linked lists.
- **Non-linear:** elements are not arranged sequentially. Examples: trees and graphs.

Example of a linear structure: `10 → 20 → 30 → 40 → 50`

Example of a non-linear structure:
```text
    A
   / \
  B   C
     / \
    D   E
```

## What is an Array?
An array stores multiple values of the same data type in a sequence. Python's `array` module uses a type code to specify the element type. Unlike arrays, Python lists can contain values of mixed types.

Common array operations include traversal, insertion, deletion, searching, updating, counting, reversing, extending, and sorting.

### Program 10: Bubble Sort
Bubble Sort repeatedly compares adjacent elements and swaps them when they are in the wrong order. After each pass, the largest unsorted element moves to the end. If a pass makes no swaps, the algorithm stops early.

**How it works**
1. `for i in range(n)` runs passes through the array.
2. `swapped = False` tracks whether a swap occurs during a pass.
3. `for j in range(0, n - i - 1)` visits the unsorted portion.
4. If `arr[j] > arr[j + 1]`, the adjacent elements are out of order.
5. `arr[j], arr[j + 1] = arr[j + 1], arr[j]` swaps them.
6. If no swap occurred, `break` stops the algorithm early.

````python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break

arr = [64, 34, 25, 12, 22, 11, 90]
bubbleSort(arr)
print(arr)
````

**Expected output**
```text
[11, 12, 22, 25, 34, 64, 90]
```

### Program 11: Insertion Sort
Insertion Sort builds a sorted portion of the array by taking each next element and inserting it into its correct position. It starts at the second element because the first element alone is already sorted.

**How it works**
1. Start from the second element.
2. Store the current element in `key`.
3. Shift larger elements one position to the right.
4. Insert `key` into its correct position.
5. Repeat until the entire array is sorted.

**Code explanation**
- `for i in range(1, n)` visits each element after the first.
- `key = arr[i]` stores the element being inserted.
- The inner `while` loop shifts larger elements to the right.
- `arr[j + 1] = key` inserts the element in its sorted position.

````python
def insertionSort(arr):
    n = len(arr)
    if n <= 1:
        return
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

arr = [12, 11, 13, 5, -1]
insertionSort(arr)
print(arr)
````

**Expected output**
```text
[-1, 5, 11, 12, 13]
```
