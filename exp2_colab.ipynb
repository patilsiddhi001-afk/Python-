```python
# 7. Counting Frequency of Elements
arr = [1, 2, 2, 3, 1, 4, 2]
freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

print(freq)
```

```python
# DSA-Style Example: Find Largest of Three Numbers
a, b, c = 10, 25, 15

if a > b:
    if a > c:
        print("Largest:", a)
    else:
        print("Largest:", c)
else:
    if b > c:
        print("Largest:", b)
    else:
        print("Largest:", c)
```

```python
# Nested Loop Example
for i in range(1, 4):  # Outer loop for rows
    for j in range(1, 4):  # Inner loop for columns
        print(j, end=" ")
    print()
```

```python
# Program 1: Right-Angled Triangle
for i in range(1, 6):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

```python
# Program 2: Inverted Right-Angled Triangle
for i in range(5, 0, -1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```

```python
# Program 3: Number Triangle Pattern
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# Program 4: Pyramid Pattern
rows = 5

for i in range(1, rows + 1):
    print(" " * (rows - i), end="")
    print("* " * i)
```

```python
# Program 6: Alphabet Triangle
ch = 65  # ASCII for 'A'

for i in range(5):
    for j in range(i + 1):
        print(chr(ch), end=" ")
        ch += 1
    print()
```

```python
# Program 7: Floyd's Triangle
num = 1

for i in range(1, 6):
    for j in range(1, i + 1):
        print(num, end=" ")
        num += 1
    print()
```

```python
# Program 8: Diamond Pattern
rows = 5

# Upper part
for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)

# Lower part
for i in range(rows - 1, 0, -1):
    print(" " * (rows - i) + "* " * i)
```

```python
# Continue Statement Example
for num in range(1, 11):
    if num % 3 == 0:
        continue  # Skip multiples of 3
    print(num)
```
