## Introduction to NumPy

NumPy is a core Python library for scientific computing. It provides a high-performance multidimensional array object and tools for working with arrays. NumPy arrays usually contain elements of the same data type.

### Program 1 — Create and modify a one-dimensional array

```python
import numpy as np

# 1D array
a = np.array([1, 2, 3])
print("Type:", type(a))
print("Shape:", a.shape)
print("Elements:", a[0], a[1], a[2])

a[0] = 5
print("Modified array:", a)
```

### Program 2 — Create an array of zeros using `np.zeros()`

```python
import numpy as np

zeros_array = np.zeros((3, 3))
print(zeros_array)
```

### Program 3 — Create an array of ones using `np.ones()`

```python
import numpy as np

ones_array = np.ones((3, 2))
print(ones_array)
```

### Program 4 — Generate values using `np.arange()`

```python
import numpy as np

range_array = np.arange(0, 10, 2)
print(range_array)
```

### Program 5 — Generate evenly spaced values using `np.linspace()`

```python
import numpy as np

linear_array = np.linspace(0, 1, 5)
print(linear_array)
```

### Program 6 — Create and access a two-dimensional array

```python
import numpy as np

b = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape:", b.shape)
print("Selected elements:", b[0, 0], b[0, 1], b[1, 0])
print("Array:\n", b)
```

### Program 7 — Generate random values using `np.random.rand()`

```python
import numpy as np

random_array = np.random.rand(2, 3)
print(random_array)
```

### Program 8 — Array addition, sum, square root and transpose

```python
import numpy as np

arr1 = np.array([[4, 7], [2, 6]], dtype=np.float64)
arr2 = np.array([[3, 6], [2, 8]], dtype=np.float64)

print("Array 1:\n", arr1)
print("Array 2:\n", arr2)
print("Addition:\n", np.add(arr1, arr2))
print("Sum of arr1:", np.sum(arr1))
print("Square root of arr1:\n", np.sqrt(arr1))
print("Transpose of arr1:\n", arr1.T)
```

### Program 9 — Arithmetic operations on NumPy arrays

```python
import numpy as np

x = np.array([[1, 2], [3, 4]], dtype=np.float64)
y = np.array([[5, 6], [7, 8]], dtype=np.float64)

print("Addition using +:\n", x + y)
print("Addition using np.add():\n", np.add(x, y))
print()
print("Subtraction using -:\n", x - y)
print("Subtraction using np.subtract():\n", np.subtract(x, y))
print()
print("Multiplication using *:\n", x * y)
print("Multiplication using np.multiply():\n", np.multiply(x, y))
print()
print("Division using /:\n", x / y)
print("Division using np.divide():\n", np.divide(x, y))
print()
print("Square root of x:\n", np.sqrt(x))
```

### Program 10 — Dot product of vectors and matrix-vector multiplication

```python
import numpy as np

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])
v = np.array([9, 10])
w = np.array([11, 12])

print("Matrix x:\n", x)
print("Matrix y:\n", y)
print("Vector v:", v)
print("Vector w:", w)

# Inner product: (9 * 11) + (10 * 12) = 219
print("v.dot(w):", v.dot(w))
print("np.dot(v, w):", np.dot(v, w))
print("Element-wise multiplication:", np.multiply(v, w))

print("x.dot(v):\n", x.dot(v))
print("np.dot(x, v):\n", np.dot(x, v))
```

### Program 11 — Calculate the mean using `np.mean()`

```python
import numpy as np

array = np.array([1, 2, 3])
mean_value = np.mean(array)
print("Mean:", mean_value)
```

### Program 12 — Find maximum and minimum values

```python
import numpy as np

array = np.array([1, 2, 3])
print("Array:", array)

max_val = np.max(array)
min_val = np.min(array)

print("Maximum:", max_val)
print("Minimum:", min_val)
```

### Program 13 — Access an array element using indexing

```python
import numpy as np

array = np.array([1, 2, 3, 4])
element = array[2]
print("Element at index 2:", element)
```

### Program 14 — Extract a subset using slicing

```python
import numpy as np

array = np.array([1, 2, 3, 4])
subset = array[1:3]
print("Subset:", subset)
```

### Program 15 — Sort an array using `np.sort()`

```python
import numpy as np

sorted_array = np.sort(np.array([3, 1, 2]))
print("Sorted array:", sorted_array)
```

### Program 16 — Find indices using `np.where()`

```python
import numpy as np

array = np.array([1, 2])
indices = np.where(array > 2)
print("Indices where array > 2:", indices)
```

## Experiment 7 — Quick recap

- `np.array()` creates an array.
- `np.zeros()`, `np.ones()`, `np.arange()` and `np.linspace()` create arrays in different ways.
- NumPy supports arithmetic, aggregation (`sum`, `mean`, `max`, `min`), square roots, transpose and dot products.
- Indexing selects individual elements; slicing selects a range; `np.sort()` sorts values; `np.where()` finds indices matching a condition.
