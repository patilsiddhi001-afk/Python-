## Creating a Set

```python
numbers={10,20,30,40,50}
print("Set:",numbers)
print("Type:",type(numbers))
```

## Using set()

```python
numbers=set([10,20,30,40,50])
print(numbers)
letters=set("PYTHON")
print(letters)
```

## Accessing Set Elements

```python
numbers={10,20,30,40,50}
print(30 in numbers)
print(100 in numbers)
for x in numbers:
    print(x)
```

## Adding and Updating

```python
numbers={10,20,30}
print(numbers)
numbers.add(40)
print(numbers)
numbers.update([50,60,70])
print(numbers)
```

## Deleting Set Elements

```python
numbers={10,20,30,40,50}
print(numbers)
numbers.remove(30)
print(numbers)
numbers.discard(40)
print(numbers)
numbers.pop()
print(numbers)
numbers.clear()
print(numbers)
```

## Creating Dictionary

```python
student={"name":"Rahul","roll_no":101,"branch":"CSE","percentage":85.5}
print(student)
print(type(student))
```

## Access Dictionary

```python
student={"name":"Rahul","roll_no":101,"branch":"CSE","percentage":85.5}
print(student["name"])
print(student["roll_no"])
print(student.get("branch"))
print(student.get("percentage"))
```

## Update Dictionary

```python
student={"name":"Rahul","roll_no":101,"branch":"CSE","percentage":85.5}
student["percentage"]=90.5
student["city"]="Pune"
print(student)
```

## Delete Dictionary

```python
student={"name":"Rahul","roll_no":101,"branch":"CSE","percentage":85.5}
del student["branch"]
print(student)
student.pop("percentage")
print(student)
student.popitem()
print(student)
student.clear()
print(student)
```

## Loop Dictionary

```python
student={"name":"Rahul","roll_no":101,"branch":"CSE","percentage":85.5}
for k in student: print(k)
print()
for v in student.values(): print(v)
print()
for k,v in student.items(): print(k,":",v)
```

## Student Marks

```python
marks={"Python":85,"Data Structure":90,"Database":78,"Computer Network":88}
for s,m in marks.items():
    print(s,":",m)
print("Total Marks:",sum(marks.values()))
print("Highest Marks:",max(marks.values()))
```

## Set Operations

```python
set1={10,20,30,40}
set2={30,40,50,60}
print("Union:",set1|set2)
print("Intersection:",set1&set2)
print("Difference:",set1-set2)
print("Symmetric Difference:",set1^set2)
```
