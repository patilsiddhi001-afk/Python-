## Object Identity

```python
a=12
print(id(a))
b=a
print(id(b))
a=13
print(id(a))
```

## Creating Lists

```python
a=[1,2,3,4,5]
b=['apple','banana','cherry']
c=[1,'hello',3.14,True]
print(a);print(b);print(c)
a=[10,20,"GfG",40,True]
print(a)
print(a[0]);print(a[1]);print(a[2])
```

## List Constructor

```python
a=list((1,2,3,'apple',4.5))
print(a)
b=list("MIT")
print(b)
```

## Repeated Elements

```python
a=[2]*5
b=[0]*7
print(a)
print(b)
```

## Accessing Elements

```python
a=[10,20,30,40,50]
print(a[0]);print(a[4]);print(a[-1]);print(a[1:4])
```

## Adding Elements

```python
a=[]
a.append(10)
print(a)
a.insert(0,5)
print(a)
a.extend([15,20,25])
print(a)
a.clear()
print(a)
```

## Appending Loop

```python
a=[]
for i in range(5):
    a.append(i)
print(a)
```

## Concatenate Lists

```python
list1=[1,2,3]
list2=[4,5,6]
list1=list1+list2
print(list1)
```

## Tuple in List

```python
list1=[1,2,3,4,5,6]
num_tuple=(4,5,6)
list1.insert(2,num_tuple)
print(list1)
print(list1[2][1])
```
